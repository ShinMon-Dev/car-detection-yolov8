## 2) Data Training

```python
from ultralytics import YOLO

# Load a pretrained YOLOv8 model
model = YOLO('yolov8n.pt') 

# Train the model
results = model.train(
    data=r"E:\Stanford-Cars-Dataset\filtered_dataset\dataset.yaml",      # replace with the correct path to your dataset.yaml
    epochs=100,                                 # number of epochs
    imgsz=640,                                  # image size
    batch=16,                                   # adjust based on your GPU memory
    optimizer='SGD',                            # SGD optimizer (you can change to Adam if you want)
    hsv_h=0.015,                                # data augmentation: hue
    hsv_s=0.7,                                  # data augmentation: saturation
    hsv_v=0.4,                                  # data augmentation: value
    degrees=10,                                 # data augmentation: rotation
    translate=0.1,                              # data augmentation: translation
    scale=0.5,                                  # data augmentation: scaling
    shear=2.0,                                  # data augmentation: shearing
    perspective=0.0005,                         # slight perspective distortion
    flipud=0.0,                                 # vertical flip (disabled)
    fliplr=0.5,                                 # horizontal flip (enabled)
    mosaic=1.0,                                 # mosaic augmentation (enabled)
    mixup=0.0,                                  # mixup augmentation (disabled)
    cache=True,                                 # cache images for faster training
    workers=4,                                  # number of CPU workers
    device=0                                    # use GPU 0
)
```
